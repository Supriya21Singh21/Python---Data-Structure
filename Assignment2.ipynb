8.  What is a hash table, and how does it relate to dictionaries in Python?

  - A hash table is like a super-fast locker system.

     -  In Python, a dictionary uses this system to store things.

     -  You give it a key (like a name), and it uses a special math trick (called a hash function) to figure out where to keep the value.

     -  Later, when you ask for that key again, it quickly finds the value in the same spot.

       Example:
                Think of a dictionary like a row of lockers:

     You say: "Put 'Supriya' in locker for 'name'."

     Python uses the hash table to find the right locker for "name" and stores "Supriya" there.

     When you ask for "name" again, it knows exactly which locker to open — super fast!

29.  Write a code to combine two dictionaries into one and print the result.

```python
# Combine two dictionaries into one and print the result

dict1 = {"name": "Alice", "age": 30}
dict2 = {"city": "Paris", "country": "France"}

# Method 1: Using the unpacking operator (Python 3.5+)
combined_dict = {**dict1, **dict2}
print("Combined dictionary:", combined_dict)

# Method 2: Using .update() (modifies the first dictionary)
dict1_copy = dict1.copy()
dict1_copy.update(dict2)
print("Combined using .update():", dict1_copy)
```
